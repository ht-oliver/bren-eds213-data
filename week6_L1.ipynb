## Using a cursor

```python
cur = conn.cursor()
```

```python
cur.execute("SELECT * FROM Site")
```

```python
cur.fetchone()
```

```python
cur.fetchone()[0]
```

```python
# We want to iterate through the rows and print the site code
cur.execute("SELECT * FROM Site")

for row in cur.fetchall():
    print(row[0])
```

```python
cur.execute("SELECT * from Site")

while True:
    row = cur.fetchone()
    if row == None:
        break
    print(f"Site code: {row[0]}")
```

```python
# you can also create table
# If you want a block of quotes in python, use three ' "''s
cur.execute("""CREATE TEMP TABLE temp_nests AS
            SELECT * FROM Bird_nests""")
```
